Energy balance

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

def plot_network_panel(
    n,
    time_slice,
    tech_mapping,
    load_mapping,
    color_series,
    label_series,
    title="Network",
    ylim=(-220, 220),
    price_ylim=(-300,300),
    font_size=7,  # Reduced from 7
    ax=None
):
    """
    Plot a single network panel with stacked energy (supply/demand) and optional hourly price.
    Returns ax and ax2 (secondary y-axis for price), with zero-line aligned.
    """
    if ax is None:
        fig, ax = plt.subplots(figsize=(6,4))
    else:
        fig = ax.figure

    # --- Prepare energy data ---
    p = (
        n.statistics.energy_balance(aggregate_time=False, bus_carrier="AC", nice_names=False)
        .dropna(how="all")
        .groupby("carrier")
        .sum()
        .div(1e3)
        .T
    )
    p.rename(columns={"": "Load"}, inplace=True)
    p_week = p.loc[time_slice]
    p_week.index = p_week.index.map(lambda dt: dt.replace(year=2045))

    p_positive = p_week.where(p_week >= 0).dropna(how="all", axis=1)
    p_negative = p_week.where(p_week < 0).dropna(how="all", axis=1)

    p_positive.columns = p_positive.columns.map(tech_mapping)
    p_positive = p_positive.groupby(p_positive.columns, axis=1).sum()
    p_positive = p_positive.reindex(columns=reversed(color_series.keys()), fill_value=0)
    p_positive = p_positive.drop(columns=["load"], errors='ignore')

    p_negative.columns = p_negative.columns.map(load_mapping)
    p_negative = p_negative.reindex(columns=reversed(color_series.keys()), fill_value=0)

    # --- Plot energy ---
    p_positive.plot.area(
        stacked=True,
        color=[color_series[col] for col in p_positive.columns],
        linewidth=0,
        legend=False,
        ax=ax
    )
    p_negative.plot.area(
        stacked=True,
        color=[color_series[col] for col in p_negative.columns],
        linewidth=0,
        legend=False,
        ax=ax
    )

    ax.set_title(title, fontsize=font_size, pad=2)  # Reduced title padding
    ax.set_xlabel("")
    ax.set_ylabel("Electricity (GW)", fontsize=font_size)
    ax.set_ylim(*ylim)
    ax.tick_params(axis='both', labelsize=font_size, which='major', pad=1)  # Reduced tick padding
    ax.tick_params(axis='x', labelsize=font_size, which='both', pad=1)
    
    # Reduce margins around the plot
    ax.margins(x=0.01, y=0.02)
    
    # --- Prepare price ---
    lmp = n.buses_t.marginal_price.loc[:, n.buses["carrier"] == "AC"]
    load = n.buses_t.marginal_price.loc[:, n.buses["carrier"] == "AC"]
    hourly_price = (load * lmp).sum(axis=1) / load.sum(axis=1)
    hourly_price = hourly_price.loc[time_slice]
    hourly_price.index = p_week.index
    hourly_price = hourly_price.to_frame(name='Price')

    # --- Plot price ---
    ax2 = ax.twinx()
    hourly_price.plot(
        y='Price',
        color='red',
        linestyle='-',
        linewidth=0.5,
        ax=ax2,
        legend=False
    )
    ax2.set_ylabel("Price (US$/MWh)", fontsize=font_size, color='red')
    ax2.tick_params(axis='y', labelsize=font_size, colors='red', pad=1)  # Reduced tick padding

    ax2.yaxis.set_label_coords(1.1, 0.75)  # x=1.1 (right of axis), y=0.75 (75% up from bottom)

    # --- Apply manual price limits if provided ---
    if price_ylim:
        ax2.set_ylim(*price_ylim)
        ax2.set_autoscale_on(False)
        ax2.set_yticks([t for t in ax2.get_yticks() if 0 <= t <= price_ylim[1]])

    # --- Create legend ---
    def make_patches(df, cols):
        patches = []
        for col in cols:
            if df[col].abs().sum() > 0:
                label = label_series.get(col, col)
                patches.append(mpatches.Patch(color=color_series[col], label=label))
        return patches

    supply_patches = make_patches(p_positive, p_positive.columns)[::-1]
    demand_patches = make_patches(p_negative, p_negative.columns)
    supply_header = mpatches.Patch(color='none', label='Supply')
    demand_header = mpatches.Patch(color='none', label='Demand')

    all_handles = [supply_header] + supply_patches + [demand_header] + demand_patches
    legend = ax.figure.legend(
        handles=all_handles,
        loc='center left',
        bbox_to_anchor=(1, 0.5),  # Moved closer to plots
        frameon=False,
        handlelength=1.2,  # Reduced from 1.5
        handletextpad=0.3,  # Reduced from 0.5
        labelspacing=0.2,   # Reduced from 0.3
        fontsize=font_size,
        columnspacing=0.5   # Add column spacing control if using multiple columns
    )
    for text in legend.get_texts():
        if text.get_text() in ("Supply", "Demand"):
            text.set_fontweight('bold')

    return ax, ax2


# Create denser subplot layout
fig, axes = plt.subplots(1, 2, figsize=(7, 3), sharey=True)  # Reduced figure size
plt.subplots_adjust(wspace=0.15, hspace=0.1)  # Explicit spacing control

time_slice = slice('2013-07-15', '2013-07-21')

ax_flex, ax2_flex = plot_network_panel(
    n_flex, time_slice, tech_mapping, load_mapping, color_series, label_series,
    title="With demand-side flexibility", ax=axes[0], ylim=(-220, 220), price_ylim=(-250, 250)
)

ax_noflex, ax2_noflex = plot_network_panel(
    n_noflex, time_slice, tech_mapping, load_mapping, color_series, label_series,
    title="Without demand-side flexibility", ax=axes[1], ylim=(-220, 220), price_ylim=(-250, 250)
)

# Use constrained_layout for better automatic spacing
plt.tight_layout(pad=0.5)  # Reduced padding
plt.savefig("energy_balance_summer.png", dpi=300, bbox_inches='tight', pad_inches=0.05)  # Reduced padding
plt.show()


# Winter plot with same dense settings
time_slice = slice('2013-01-14', '2013-01-20')

fig, axes = plt.subplots(1, 2, figsize=(7, 3), sharey=True)  # Reduced figure size
plt.subplots_adjust(wspace=0.15, hspace=0.1)

ax_flex, ax2_flex = plot_network_panel(
    n_flex, time_slice, tech_mapping, load_mapping, color_series, label_series,
    title="With demand-side flexibility", ax=axes[0], ylim=(-280, 280), price_ylim=(-800,800)
)
ax_noflex, ax2_noflex = plot_network_panel(
    n_noflex, time_slice, tech_mapping, load_mapping, color_series, label_series,
    title="Without demand-side flexibility", ax=axes[1], ylim=(-280, 280), price_ylim=(-800, 800)
)

plt.tight_layout(pad=0.5)
plt.savefig("energy_balance_winter.png", dpi=300, bbox_inches='tight', pad_inches=0.05)
plt.show()
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

# Modify your function to accept an axis
def plot_pypsa_network_ax(network, ax, tech_mapping=None, color_series=None, 
                         bus_scale=3e8, line_scale=2e3,
                         extent=[5.8, 15.1, 47.2, 55.2]):
    # Create a copy of the network to avoid modifying the original
    n = network.copy()
    
    # Clean up the network data
    if '' in n.carriers.index:
        n.carriers = n.carriers.drop(index='')
    
    # Remove load shedding from generators
    load_shedding_mask = n.generators.filter(like="load", axis="rows").index
    if len(load_shedding_mask) > 0:
        n.generators = n.generators.drop(index=load_shedding_mask)
    
    # Remove H2 transfer to H2 demand REMIND links
    h2_demand_mask = n.links.filter(like="H2 demand", axis="rows").index
    if len(h2_demand_mask) > 0:
        n.links = n.links.drop(index=h2_demand_mask)
    
    # Calculate annual generation from generators
    buses = n.generators_t.p.sum().groupby([n.generators.bus, n.generators.carrier]).sum()
    
    # Calculate battery discharge
    battery_links = n.links.carrier == "battery discharger"
    if battery_links.any():
        battery = n.links_t.p0.loc[:, battery_links].sum()
        battery.index = battery.index.map(n.links.bus1)
        battery_multiindex = pd.MultiIndex.from_tuples(
            [(bus, 'battery discharger') for bus in battery.index],
            names=['bus', 'carrier']
        )
        battery_series = pd.Series(battery.values, index=battery_multiindex)
    else:
        battery_series = pd.Series(dtype=float)
    
    # Calculate fuel cell generation
    fuelcell_links = n.links.carrier == "H2 fuel cell"
    if fuelcell_links.any():
        fuelcell = n.links_t.p0.loc[:, fuelcell_links].sum()
        fuelcell.index = fuelcell.index.map(n.links.bus1)
        fuelcell_multiindex = pd.MultiIndex.from_tuples(
            [(bus, 'H2 fuel cell') for bus in fuelcell.index],
            names=['bus', 'carrier']
        )
        fuelcell_series = pd.Series(fuelcell.values, index=fuelcell_multiindex)
    else:
        fuelcell_series = pd.Series(dtype=float)
    
    # Combine all generation data
    combined = pd.concat([buses, battery_series, fuelcell_series])
    
    # Set custom colors for carriers if mappings are provided
    if tech_mapping is not None and color_series is not None:
        for carrier in n.carriers.index:
            tech = tech_mapping.get(carrier, carrier)
            n.carriers.loc[carrier, 'color'] = color_series.get(tech, 'gray')
    
    # Create the network plot
    n.plot.map(ax=ax,
               bus_sizes=combined / bus_scale,
               bus_alpha=0.8,
               line_widths=n.lines.s_nom_opt / line_scale,
               link_widths=n.links.p_nom_opt / line_scale,
               color_geomap=False)
    
    # Add geographic features
    ax.add_feature(cfeature.OCEAN, color='lightblue', alpha=0.5)
    ax.add_feature(cfeature.LAND, color='lightgray', alpha=0.3)
    ax.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor='black', alpha=0.6)
    ax.add_feature(cfeature.COASTLINE, linewidth=0.5, edgecolor='black', alpha=0.4)
    
    # Set extent
    ax.set_extent(extent, crs=ccrs.PlateCarree())
    
    return ax

# Set global font size
plt.rcParams.update({'font.size': 7})

# Create single figure with two panels
fig, axes = plt.subplots(1, 2, figsize=(7, 6), 
                        subplot_kw={'projection': ccrs.PlateCarree()})

# Plot the 2030 networks
# plot_pypsa_network_ax(n_flex_2030, axes[0, 0], tech_mapping, color_series)
# axes[0, 0].set_title('With demand-side flexibility (2030)', fontsize=7)

# plot_pypsa_network_ax(n_noflex_2030, axes[0, 1], tech_mapping, color_series)
# axes[0, 1].set_title('Without demand-side flexibility (2030)', fontsize=7)

# Plot the 2045 networks
plot_pypsa_network_ax(n_flex, axes[0], tech_mapping, color_series)
axes[0].set_title('With demand-side flexibility (2045)', fontsize=7)

plot_pypsa_network_ax(n_noflex, axes[1], tech_mapping, color_series)
axes[1].set_title('Without demand-side flexibility (2045)', fontsize=7)

plt.tight_layout()
plt.savefig("pypsa_networks.png", dpi=300, bbox_inches='tight', pad_inches=0.05)
plt.show()
```
